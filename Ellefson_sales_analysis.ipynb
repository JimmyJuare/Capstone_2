### What is monthly total revenue for in-store sales in each of the two sales territories, over the full period covered by the data?

```python
curr_store_sales = store_sales.copy()
```

```python

# Convert date and create Month column
curr_store_sales['Transaction Date'] = pd.to_datetime(curr_store_sales['Transaction Date'])
curr_store_sales['Year-Month'] = curr_store_sales['Transaction Date'].dt.to_period('M')   # e.g., 2022-01

merged_sales = pd.merge(
    left=curr_store_sales,           
    right=store_detail,           
    on='Store ID',              # Common column between both tables
    how='left'                  # Keep all sales records
)
# Group by Month and sum the Sale Amount
monthly_revenue = (merged_sales
                   .groupby(['Year-Month', 'Territory Manager'])['Sale Amount']
                   .sum()
                   .reset_index())

monthly_revenue = monthly_revenue.rename(columns={'Sale Amount': 'Monthly Revenue'})

# Filter for the two territories/managers
plot_data = monthly_revenue[
    (monthly_revenue['Territory Manager'] == 'See Ellefson') | 
    (monthly_revenue['Territory Manager'] == 'Jim Heck')
]

# Pivot the data so each manager becomes a column (better for plotting)
pivot_data = plot_data.pivot(index='Year-Month', 
                             columns='Territory Manager', 
                             values='Monthly Revenue')

# Convert Period to string for clean x-axis labels
pivot_data.index = pivot_data.index.astype(str)

# Create the Line Chart
plt.figure(figsize=(12, 7))

for manager in pivot_data.columns:
    plt.plot(pivot_data.index, 
             pivot_data[manager], 
             marker='o', 
             linewidth=2.5,
             label=manager)

plt.title('Monthly Total Revenue by Sales Territory Manager', fontsize=16, pad=20)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Total Revenue ($)', fontsize=12)
plt.legend(title='Territory Manager', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.7)

# Rotate x labels for better readability
plt.xticks(rotation=45, ha='right')

# Format y-axis with dollar signs and commas
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'${x:,.0f}'))

plt.tight_layout()
plt.show()
```

 How would you rank the sales performance of each store in each sales territory? 

### Store Sales Performance Ranking

I ranked the stores by **Total Revenue** within each sales territory. Here are the key findings:

- **See Ellefson’s Territory**: The top-performing store is Store ID `850` in `New York` with $621,565.02 in total revenue.
- **Jim Heck’s Territory**: The top-performing store is Store ID `705` in `Denver` with $917,471.53 in total revenue.

**Observations**:
- The top 3 stores in each territory account for 33.78% of total revenue.
- There is a significant performance gap between the top and bottom stores.

```python
# Filter only the two relevant territory managers
two_territories = merged_sales[
    (merged_sales['Territory Manager'] == 'See Ellefson') | 
    (merged_sales['Territory Manager'] == 'Jim Heck')
]

# Total sales per store within these two territories
store_performance = (two_territories
    .groupby(['Territory Manager', 'Store ID', 'Store Location'])['Sale Amount']
    .sum()
    .reset_index()
    .rename(columns={'Sale Amount': 'Total Revenue'})
    .sort_values(['Territory Manager', 'Total Revenue'], ascending=[True, False])
)

# Add rank within each territory
store_performance['Rank'] = store_performance.groupby('Territory Manager')['Total Revenue'].rank(
    ascending=False, 
    method='dense'
).astype(int)

# Reorder columns for better readability
store_performance = store_performance[['Rank', 'Territory Manager', 'Store ID', 'Store Location', 'Total Revenue']]

# Display the final table
print("Sales Performance Ranking by Territory (See Ellefson & Jim Heck)")
display(
    store_performance.style
    .format({'Total Revenue': '${:,.2f}'})
    .set_caption("Store Sales Performance Ranking - Two Territories")
    .hide(axis='index')   # Hides the default row numbers
    
)
```

Comparing the customer ID from the customer list data with the rewards ID from the sales data,
who were the top customers in each sales territory?

```python
# merge customer list with sales data using Customer ID = Rewards ID
customer_sales = pd.merge(
    left=merged_sales,                    # Your sales + store data
    right=customer_list,                      # Customer list dataframe
    left_on='RewardsID',                 # From sales
    right_on='Customer ID',               # From customer list
    how='left'
)

# Calculate total spending per customer in each territory
top_customers = (customer_sales
    .groupby(['Territory Manager', 'Customer ID', 'Customer Name'])['Sale Amount']
    .sum()
    .reset_index()
    .rename(columns={'Sale Amount': 'Total Spent'})
    .sort_values(['Territory Manager', 'Total Spent'], ascending=[True, False])
)

# Add ranking within each territory
top_customers['Rank'] = top_customers.groupby('Territory Manager')['Total Spent'].rank(
    ascending=False, 
    method='dense'
).astype(int)

# Keep only the two territories
top_customers = top_customers[
    (top_customers['Territory Manager'] == 'See Ellefson') | 
    (top_customers['Territory Manager'] == 'Jim Heck')
]

# Reorder columns nicely
top_customers = top_customers[['Rank', 'Territory Manager', 'Customer ID', 
                               'Customer Name', 'Total Spent']]

# Show only top 5 customers per territory
top_5_per_territory = top_customers.groupby('Territory Manager').head(5)

display(top_5_per_territory.style
        .format({'Total Spent': '${:,.2f}', 'Customer ID': '{:.0f}'})
        .set_caption("Top 5 Customers per Territory").hide(axis='index'))
```
