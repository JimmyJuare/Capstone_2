### What is monthly total revenue for in-store sales in each of the two sales territories, over the full period covered by the data?

```python
display(store_detail)
```

```python
curr_store_sales = store_sales.copy()
```

```python

# Convert date and create Month column
curr_store_sales['Transaction Date'] = pd.to_datetime(curr_store_sales['Transaction Date'])
curr_store_sales['Month'] = curr_store_sales['Transaction Date'].dt.to_period('M')   # e.g., 2022-01
curr_store_sales['Year'] = curr_store_sales['Transaction Date'].dt.year

# Group by Month and sum the Sale Amount
monthly_revenue = (curr_store_sales
                   .groupby('Month')['Sale Amount']
                   .sum()
                   .reset_index())

monthly_revenue = monthly_revenue.rename(columns={'Sale Amount': 'Monthly Revenue'})

print("Monthly Revenue Across All Months:")
display(monthly_revenue)
```

```python
# lets start with merging the store_sales and store_detail tables to get a better understanding 
# of the sales data in relation to the store information. We will use a left join to keep all sales 
# records and bring in the relevant store details.
merged_sales = pd.merge(
    left=store_sales,           
    right=store_detail,           
    on='Store ID',              # Common column between both tables
    how='left'                  # Keep all sales records
)
print("\nColumns now available:")
print(merged_sales.columns.tolist())
```
